```python
#Create List
Student_list = ["Aditya", "Ayush", "Aneeket", "Yash", "Darsh"]

print("\nStudent List:")
print(Student_list)

#Create Tuple
Student_tuple = ("Aditya", "Ayush", "Aneeket", "Yash", "Darsh")

print("\nStudent Tuple:")
print(Student_tuple)

#Create Dictionary
Student_dict = {
 1:"Aditya",
 2:"Ayush" ,
 3:"Aneeket",
 4:"Yash",
 5:"Darsh"  
    }
print("\nStudent Dictionary:")
print(Student_dict)

#Adding in Dictionary
Student_dict[6]="Virag"

print("\nUpdated Student Dictionary:")
print(Student_dict)

#Replaceing names
Student_dict[2]="Kunal"

print("\nUpdated Student Dictionary:")
print(Student_dict)

#Delete a student
del Student_dict[4]

print("\nUpdated Student Dictionary:")
print(Student_dict)
```
