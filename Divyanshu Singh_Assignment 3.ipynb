```python
#To check whether or not the triangle is a right-angled triangle 

def is_right_triangle():
    a=float(input("Enter the length of the first side: "))
    b=float(input("Enter the length of the second side: "))
    c=float(input("Enter the length of the third side: "))

    #store them in a list and sort them in ascending order
    sides = [a, b, c]
    sides.sort()

    #using pythagoras theorem
    if sides [0]**2 + sides [1]**2 == sides [2]**2:
        print("This triangle is a right-angled triangle")
    else:
        print("This triangle is not a right-angled triangle")

is_right_triangle()
```
